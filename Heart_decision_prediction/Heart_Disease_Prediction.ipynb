## **7. Feature Selection**

### **🎯 Objective**
The objective of feature selection is to identify the most relevant variables that contribute significantly to predicting heart disease while eliminating unnecessary features.

### **🔍 Observation**
Relevant predictor variables were selected based on their importance and relationship with the target variable.

### **✅ Conclusion**
Using only important features improves model performance, reduces computational complexity, and minimizes the risk of overfitting.

---

```python
# ==========================================
# Feature Selection
# ==========================================

X = df.drop("target", axis=1)

y = df["target"]
```

```python
print(X.head())
```

```python
print(y.head())
```

## **8. Train-Test Split**

### **🎯 Objective**
The objective of this step is to divide the dataset into training and testing sets so that the model can be trained on one portion of the data and evaluated on unseen data.

### **🔍 Observation**
The dataset was successfully split into training and testing datasets using an appropriate ratio.

### **✅ Conclusion**
Separating training and testing data ensures unbiased evaluation of machine learning models.

---

```python
# ==========================================
# Train-Test Split
# ==========================================
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
print("Training Data :", X_train.shape)
print("Testing Data :", X_test.shape)
```

## **9. Feature Scaling**

### **🎯 Objective**
The objective of feature scaling is to standardize numerical features so that they have similar ranges, improving the performance of distance-based algorithms.

### **🔍 Observation**
The selected numerical features were successfully transformed using a scaling technique without affecting their relationships.

### **✅ Conclusion**
Feature scaling improves the efficiency and accuracy of algorithms such as KNN, SVM, and Logistic Regression.

---

```python
# ==========================================
# Feature Scaling
# ==========================================

from sklearn.preprocessing import StandardScaler
#scaler object
scaler = StandardScaler()
# training data scaler
X_train = scaler.fit_transform(X_train)

#testing data scaler
X_test = scaler.transform(X_test)
```

```python
print(X_train)
```

## **10.5 Support Vector Machine (SVM)**

### **🎯 Objective**
Train a Support Vector Machine classifier to find the optimal decision boundary between classes.

### **🔍 Observation**
The SVM model successfully identified the separating hyperplane for classification.

### **✅ Conclusion**
SVM provides strong classification performance, especially on standardized datasets.

---

```python
# ==========================================
# Model Training
# ==========================================

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC

from sklearn.metrics import accuracy_score
```

```python
# ==========================================
# logistic Regression
# ==========================================
lr_model = LogisticRegression()

lr_model.fit(X_train, y_train)

lr_prediction = lr_model.predict(X_test)

lr_accuracy = accuracy_score(y_test, lr_prediction)

print("Logistic Regression Accuracy :", lr_accuracy)
```

```python
# ==========================================
# Decision Tree Classifier
# ==========================================
dt_model = DecisionTreeClassifier(random_state=42)

dt_model.fit(X_train, y_train)

dt_prediction = dt_model.predict(X_test)

dt_accuracy = accuracy_score(y_test, dt_prediction)

print("Decision Tree Accuracy :", dt_accuracy)
```

```python
# ==========================================
# Random forest
# ==========================================

rf_model = RandomForestClassifier(random_state=42)

rf_model.fit(X_train, y_train)

rf_prediction = rf_model.predict(X_test)

rf_accuracy = accuracy_score(y_test, rf_prediction)

print("Random Forest Accuracy :", rf_accuracy)
```

```python
# ==========================================
# k-nearest neighbours
# ==========================================

knn_model = KNeighborsClassifier()

knn_model.fit(X_train, y_train)

knn_prediction = knn_model.predict(X_test)

knn_accuracy = accuracy_score(y_test, knn_prediction)

print("KNN Accuracy :", knn_accuracy)
```

```python
# ==========================================
# Support Vector Machine
# ==========================================

svm_model = SVC()

svm_model.fit(X_train, y_train)

svm_prediction = svm_model.predict(X_test)

svm_accuracy = accuracy_score(y_test, svm_prediction)

print("SVM Accuracy :", svm_accuracy)
```

## **11. Accuracy Comparison**

### **🎯 Objective**
Compare the performance of all trained machine learning models using their accuracy scores.

### **🔍 Observation**
The accuracy of each classifier was calculated and compared to identify the best-performing model.

```python
# ==========================================
# Accuracy Comparison
# ==========================================
accuracy = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest",
        "KNN",
        "Support Vector Machine"
    ],
    "Accuracy": [
        lr_accuracy,
        dt_accuracy,
        rf_accuracy,
        knn_accuracy,
        svm_accuracy
    ]
})

accuracy
```

```python
# ==========================================
# Classification Report 
# ==========================================
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
from sklearn.metrics import ConfusionMatrixDisplay

print(classification_report(y_test, rf_prediction))
```

## **13. Confusion Matrix**

### **🎯 Objective**
Analyze prediction performance by comparing actual and predicted class labels.

### **🔍 Observation**
The confusion matrix displayed True Positives, True Negatives, False Positives, and False Negatives.

### **✅ Conclusion**
The confusion matrix provides deeper insights into the types of classification errors made by the model.

---

```python
# ==========================================
# Confusion Matrix
# ==========================================
cm = confusion_matrix(y_test, rf_prediction)

display = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["No Disease", "Disease"]
)

display.plot(cmap="Blues")

plt.title("Confusion Matrix")

plt.show()
```
